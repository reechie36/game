```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('unigram_freq.csv')

df.head(5)
```

```python
df_5 = df[df['word'].str.len() == 5]

df_5.info()
```

```python
df_5
```

```python
#reindex the dataframe to have a new index starting from 0
df_5 = df_5.reset_index(drop=True)

df_5 
```

```python
# add a new column for points based on rank/index order
# wan to noramlize based on the max rank, so that the highest rank gets 1 point and the lowest rank gets 0 points based on frequency in the dataset
df_5['points'] = (df_5.index.max() - df_5.index) / df_5.index.max() 
df_5
```

```python
df_5.to_csv('unigram_freq_5.csv', index=True)
```

```python
wordle = pd.read_csv('wordle.csv')
wordle = wordle[["word", "occurrence"]]
wordle
```

```python
wordle_referenced = pd.merge(df_5, wordle, how='left', left_on='word', right_on='word')
wordle_referenced = wordle_referenced.dropna()

wordle_referenced
```

```python
wordle_referenced.to_csv('wordle_referenced.csv', index=True)
wordle_referenced = wordle_referenced.drop(columns=['occurrence'])
wordle_referenced 
```

```python
# Alphabetical order of the words in the wordle_referenced dataframe
wordle_referenced = wordle_referenced.sort_values(by='word')
wordle_referenced.to_csv('wordle_referenced.csv', index=True)
```
